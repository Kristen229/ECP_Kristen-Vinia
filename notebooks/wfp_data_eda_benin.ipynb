# **Analyse de données**

## Objectif du projet
  Dans le cadre de ce projet, il s'agit de comprendre comment fluctue le prix des aliments et éventuellement pourquoi cette fluctuation.Cette compréhension servirait à faire des prédictions sur les prochaines fluctuations éventulles afin de guider le choix des décideurs.

### Présentation de chaque colonne
- **country_id** (adm0_id) : renseigne sur l'identifiant unique du pays où la donnée a été collectée.

- **country_name** (adm0_name) : renseigne sur le nom du pays où la donnée a été collectée.

- **region_id** (adm1_id) : renseigne sur l'identifiant unique de la - région (premier niveau administratif) dans le pays.

- **region_name** (adm1_name) : renseigne sur le nom de la région dans le pays.

- **market_id** (mkt_id) : renseigne sur l'identifiant unique du marché spécifique où le prix a été relevé.

- **market_name** (mkt_name) : renseigne sur le nom du marché spécifique (ville ou lieu précis).

- **commodity_id** (cm_id) : renseigne sur l'identifiant unique du produit de base (aliment) dont le prix est enregistré.

- **commodity_name** (cm_name) : renseigne sur le nom du produit de base (ex: "Rice - Retail").

- **currency_id** (cur_id) : renseigne sur l'identifiant unique de la devise utilisée pour le prix.

- **currency_name** (cur_name) : renseigne sur le nom de la devise (ex: "CFA Franc").

- **point_type_id** (pt_id) : renseigne sur l'identifiant unique du type de point de vente (ex: gros, détail).

- **point_type_name** (pt_name) : renseigne sur le nom du type de point de vente (ex: "Wholesale", "Retail").

- **unit_id** (um_id) : renseigne sur l'identifiant unique de l'unité de mesure utilisée (ex: kg, litre).

- **unit_name** (um_name) : renseigne sur le nom de l'unité de mesure utilisée (ex: "KG", "L").

- **month** (mp_month) : renseigne sur le mois de la collecte du prix (format numérique).

- **year** (mp_year) : renseigne sur l'année de la collecte du prix.

- **price** (mp_price) : renseigne sur le prix enregistré pour le produit, l'unité et la devise spécifiés.

- **price_source** (mp_commoditysource) : renseigne sur la source spécifique ou la méthode utilisée pour obtenir le prix.

la base contient des infos sur des produits qui ne font partie d'aucune des catégories que nous étudions. cependant on a des infos sur les éréales et les huiles et on note qu'il y en a plus sur les céréales que sur les huiles

```python
# Prix moyen par  catégorie de produit
plt.figure(figsize=(8, 5))
sns.barplot(data=benin_filtered, x='product_category', y='price', palette='muted')
plt.title('Prix Moyen par Catégorie de Produit', fontweight='bold', fontsize=12)
plt.xlabel('Catégorie de Produit', fontsize=10)
plt.ylabel('Prix Moyen (XOF / KG ou L)', fontsize=10)
plt.show()
```

```python
# Version corrigée (Guillemets doubles à l'extérieur)
mean_cat_price = benin_filtered.groupby('product_category')['price'].mean()

print(f"- Le prix moyen des céréales est de {mean_cat_price['Cereals']:.0f} XOF/KG")
print(f"- Le prix moyen des huiles est de {mean_cat_price['Oils']:.0f} XOF/L")
```

- Le prix moyen au Bénin pour les produits de la catégorie Oils est significativement plus élevé que pour les Cereals.

- Le coût d'accès aux lipides (huiles) représente une charge financière bien plus importante pour les ménages que l'accès aux féculents de base (céréales).

**D'après cette analyse l'inflation sur les huiles aura un impact proportionnellement plus grand sur le budget familial.**

```python
# analyse des marges prix Gros vs Détail
plt.figure(figsize=(8, 5))
sns.barplot(data=benin_filtered, x='point_type_name', y='price', palette='coolwarm')
plt.title('Écart de Prix : Gros vs Détail', fontsize=12, fontweight='bold')
plt.ylabel('Prix Moyen (XOF)')
plt.show()
```

```python
price_diff=benin_filtered.groupby('point_type_name')['price'].mean()
marge=price_diff['Retail'] - price_diff['Wholesale']
print(f"La marge brute moyenne (différence entre le prix de Détail et de Gros) est de : {marge:.0f} XOF")
```

De l'analyse des marges, la structure du marché béninois montre une marge brute moyenne de 428 XOF entre le grossiste et le consommateur final.

**Cet écart peut être dû aux coûts logistiques et les marges des intermédiaires**

```python
# disparité régionale ; par rapport aux marchés : prix moyen par marché (tout produit confondu)
expensive_market = benin_filtered.groupby('market_name')['price'].mean().sort_values(ascending=False).head(20)

colors = sns.color_palette("Blues_r", n_colors=20)

plt.figure(figsize=(10, 6))

expensive_market.plot(kind='barh', color=colors)
plt.title('Top 20 des Marchés les plus chers au Bénin', fontsize=12, fontweight='bold')
plt.xlabel('Prix Moyen (XOF)')
plt.ylabel('Marchés')
plt.gca().invert_yaxis() # Pour avoir le plus cher en haut
plt.grid(axis='x', linestyle='--', alpha=0.7) # Ajoute une grille verticale pointillée
plt.show()
```

```python
# Top 5 des marchés les plus chers au Bénin
top_market = benin_filtered.groupby('market_name')['price'].mean().sort_values(ascending=False).head(5)

print(f"Marchés affichant les prix moyens les plus élevés : {top_market}")
```

 L'analyse géographique révèle des disparités importantes : les marchés de Comé et Savalou affichent les prix moyens les plus élevés du pays, dépassant les 800 XOF.

 Cela indique que **les populations de ces zones sont plus vulnérables à l'insécurité alimentaire liée au prix que celles des zones de production**.

**Le mois de juin est le point de vulnérabilité maximale pour le marché du maïs.**

- Le maïs est cher en Juin, mais le riz local est bas en Août. Le riz local peut servir de "tampon" ou d'alternative lorsque le maïs est inaccessible.

**Il existe un potentiel de substitution stratégique entre le maïs et le riz local.**

- Le riz importé et le blé ont des pics déconnectés des saisons béninoises (pics en août et juillet respectivement). Ces produits ne dépendent pas du climat local (sécheresse ou inondations au Bénin n'impactent pas le prix du blé mondial).

**La gestion de ces produits doit se faire via la veille des cours mondiaux et non via le calendrier agricole local.**

- L'huile de palme a un cycle différent (pic en janvier, bas en août).

**La gestion des stocks d'huile de palme doit suivre un calendrier spécifique, le mois de janvier étant le plus critique pour les prix.**

L’analyse exploratoire montre que les prix au Bénin dépendent principalement des denrées de base, en particulier les céréales, qui dominent la consommation.

Les huiles présentent des prix structurellement plus élevés que les céréales, indépendamment de leur origine, soulignant que le type de produit est un déterminant majeur du prix.

Des disparités géographiques marquées sont observées, avec des prix plus élevés sur les marchés de Comé et Savalou, et une majoration moyenne de 428 XOF par unité liée à la chaîne de distribution.

L’analyse temporelle met en évidence des saisonnalités spécifiques : le maïs est particulièrement vulnérable en juin, tandis que l’huile de palme connaît un pic de tension des prix en janvier. Des effets de substitution, notamment entre le maïs et le riz local, sont également observés.

Enfin, les chocs exogènes (crise alimentaire de 2008, fermeture des frontières avec le Nigéria, pandémie de COVID-19) ont fortement influencé les prix, en particulier ceux du maïs, du riz importé et de la farine de blé.
